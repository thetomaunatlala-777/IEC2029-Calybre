#### b. Missing Values

```python
print(df2004_renamed.isnull().sum())
```

The missing values in the "Municipality"column will be filled in by the values found in the rows with the same district number, as a ditrict number will always be under the same municipality.

```python
# lookup from rows that DO have a municipality
municipality_lookup = (
    df2004_renamed[df2004_renamed['municipality'].notna()]
    .groupby('vd_number')['municipality']
    .first()
)

# Fill missing using lookup, otherwise, replace with 'unknown'
df2004_renamed['municipality'] = df2004_renamed['municipality'].fillna(
    df2004_renamed['vd_number'].map(municipality_lookup)
).fillna('UNKNOWN')


print(f"Remaining missing: {df2004_renamed['municipality'].isna().sum()}")
```

```python
print(df2014_renamed.isnull().sum())
```

As before, we will fill in the missing values by creating a lookup table from rows where the municipality and ward number are known, mapping them to their respective voting districts. Any voting districts with no municipality or ward number on any row will be set to 'UNKNOWN'.

```python
# Fill municipality
municipality_lookup = (
    df2009_renamed[df2009_renamed['municipality'].notna()]
    .groupby('vd_number')['municipality']
    .first()
)
df2009_renamed['municipality'] = df2009_renamed['municipality'].fillna(
    df2009_renamed['vd_number'].map(municipality_lookup)
).fillna('UNKNOWN')

# Fill ward_number
ward_lookup = (
    df2009_renamed[df2009_renamed['ward_number'].notna()]
    .groupby('vd_number')['ward_number']
    .first()
)
df2009_renamed['ward_number'] = df2009_renamed['ward_number'].fillna(
    df2009_renamed['vd_number'].map(ward_lookup)
).fillna('UNKNOWN')


print(df2009_renamed[['municipality', 'ward_number']].isna().sum())
```

### 3. Demographic datasets

```python
# read and initialise tables
dfage = pd.read_sql("SELECT * FROM bronze.\"pp_age_group_27_10_2025\"", con=engine)
dfgroup = pd.read_sql("SELECT * FROM bronze.\"pp_population_group_27_10_2025\"", con=engine)
dfsex = pd.read_sql("SELECT * FROM bronze.\"pp_sex_27_10_2025\"", con=engine)
```

#### a. Naming conventions

```python

dfage_renamed = dfage.rename(columns={
    'Ward_Code': 'ward_number',
    '0-4'      : 'age_0_4',
    '5-9'      : 'age_5_9',
    '10-14'    : 'age_10_14',
    '15-19'    : 'age_15_19',
    '20-24'    : 'age_20_24',
    '25-29'    : 'age_25_29',
    '30-34'    : 'age_30_34',
    '35-39'    : 'age_35_39',
    '40-44'    : 'age_40_44',
    '45-49'    : 'age_45_49',
    '50-54'    : 'age_50_54',
    '55-59'    : 'age_55_59',
    '60-64'    : 'age_60_64',
    '65-69'    : 'age_65_69',
    '70-74'    : 'age_70_74',
    '75-79'    : 'age_75_79',
    '80-84'    : 'age_80_84',
    '85+'      : 'age_85_plus',
    'Total'    : 'total_population'
})

dfgroup_renamed = dfgroup.rename(columns={
    'Ward_Code'    : 'ward_number',
    'Black African': 'black_african',
    'Coloured'     : 'coloured',
    'Indian/Asian' : 'indian_or_asian',
    'White'        : 'white',
    'Other'        : 'other',
    'Total'        : 'total_population'
})

dfsex_renamed = dfsex.rename(columns={
    'WARD_CODE': 'ward_number',
    'Male'     : 'male',
    'Female'   : 'female',
    'Total'    : 'total_population'
})


df_population = (
    dfage_renamed
    .merge(dfgroup_renamed, on='ward_number', suffixes=('', '_groups'))
    .merge(dfsex_renamed,  on='ward_number', suffixes=('', '_sexes'))
)

# Drop duplicate total_population columns from merges
df_population = df_population.drop(columns=['total_population_groups', 'total_population_sexes'])

print(df_population.shape)
print(df_population.columns.tolist())
```

```python
# Your DataFrames + corresponding years
dfs = {
    2004: df2004_renamed,
    2009: df2009_renamed,
    2014: df2014_renamed,
    2019: df2019_renamed,
    2024: df2024_renamed
}

def standardize_basic(df, year):
    df = df.copy()


    # Add YEAR column

    df['year'] = year


    #  Ensure total_votes_cast exists
  
    if 'total_votes_cast' not in df.columns:
        df['total_votes_cast'] = (
            df.get('spoilt_votes', 0).fillna(0) +
            df.get('valid_votes', 0).fillna(0)
        )

    return df

# Apply to all datasets
dfs_standardized = {year: standardize_basic(df, year) for year, df in dfs.items()}

combined_df = pd.concat(
    dfs_standardized.values(),
    ignore_index=True,
    sort=False 
)

# Check result
print(combined_df[['year', 'total_votes_cast']].head())
```

```python
election_results_silver=combined_df
election_results_silver.head()
```

```python
mdf_silver = dfmdb_renamed
voting_stations_silver = dfvotingstations
population_silver = df_population
election_results_silver = election_results_silver

import json
from sqlalchemy.dialects.postgresql import JSONB, DOUBLE_PRECISION

dfs_silver = {
    "mdb": mdf_silver,
    "voting_stations": voting_stations_silver,
    "population": population_silver,
    "election_results": election_results_silver
}

for table_name, df in dfs_silver.items():
    df = df.copy()  

    dtype_mapping = None

    # Handle geometry columns 
    if 'geometry' in df.columns:
        df['geometry'] = df['geometry'].apply(json.dumps)
        dtype_mapping = {"geometry": JSONB}

    # Handle derived metrics 
    if table_name == "election_results":
        dtype_mapping = {
            "voter_turnout_percent": DOUBLE_PRECISION,
        }

    # Save to silver schema in chunks
    df.to_sql(
        name=table_name,
        con=engine,
        schema="silver",
        if_exists="replace",
        index=False,
        dtype=dtype_mapping,
        chunksize=50000,  
        method='multi'   
    )

    print(f"Saved {table_name} to silver layer.")
```

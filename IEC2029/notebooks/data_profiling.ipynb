### 3. Checking for duplicates

```python
SELECT vd_number, COUNT(*) AS duplicates
FROM bronze."voting_stations"
GROUP BY vd_number
HAVING COUNT(*)>1
```

```python
SELECT "WARD_CODE", COUNT(*) AS duplicates
FROM bronze."pp_sex_27_10_2025"
GROUP BY "WARD_CODE"
HAVING COUNT(*)>1
```

```python
SELECT "Ward_Code", COUNT(*) AS duplicates
FROM bronze."pp_population_group_27_10_2025"
GROUP BY "Ward_Code"
HAVING COUNT(*)>1
```

```python
SELECT "WardID", COUNT(*) AS duplicates
FROM bronze."mdb_wards_2020"
GROUP BY "WardID"
HAVING COUNT(*)>1
```

There are no duplicate records in the above datasets.

```python
SELECT "VD_Number", COUNT(*) AS duplicates
FROM bronze."2024_npe"
GROUP BY "VD_Number"
HAVING COUNT(*)>1
```

```python
SELECT "VD Number", COUNT(*) AS duplicates
FROM bronze."2019_npe"
GROUP BY "VD Number"
HAVING COUNT(*)>1
```

```python
SELECT "VOTING DISTRICT", COUNT(*) AS duplicates
FROM bronze."2014_npe"
GROUP BY "VOTING DISTRICT"
HAVING COUNT(*)>1
```

The duplicates in the NPE datasets exist due to the records showing results per political party and not just down to per voting district.

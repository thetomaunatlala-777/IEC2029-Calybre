```python
import pandas as pd
import json
from pathlib import Path
```

### Election Results data

```python
#load election results data

df2004 = pd.read_csv(r'../data/Election Results/National/2004 NPE.csv', encoding='latin1', low_memory=False, on_bad_lines='skip')
df2009 = pd.read_csv(r'../data/Election Results/National/2009 NPE.csv', encoding='latin1', low_memory=False, on_bad_lines='skip')
df2014 = pd.read_csv(r'../data/Election Results/National/2014 NPE.csv', encoding='latin1', low_memory=False, on_bad_lines='skip')
df2019 = pd.read_csv(r'../data/Election Results/National/2019 NPE.csv', encoding='latin1', low_memory=False, on_bad_lines='skip')
df2024 = pd.read_csv(r'../data/Election Results/National/2024 NPE.csv', encoding='latin1', low_memory=False, on_bad_lines='skip')
```

Above we have loaded all the election results data.

```python
# return all unique voting district numbers in order to use in api call for the voting district data
vd_nums_2004 = set(df2004['VOTING\nDISTRICT'].unique())
vd_nums_2009 = set(df2009['VOTING\nDISTRICT'].unique())
vd_nums_2014 = set(df2014['VOTING DISTRICT'].unique())
vd_nums_2019 = set(df2019['VD Number'].unique())
vd_nums_2024 = set(df2024['VD_Number'].unique())

all_vd_nums= list(set(vd_nums_2004).union(vd_nums_2009).union(vd_nums_2014).union(vd_nums_2019).union(vd_nums_2024))

all_vd_nums = sorted(int(x) for x in all_vd_nums) 
```

```python
# save the voting district numbers in a python file to be used in the api ingestion
output_file = Path("vd_nums.json")


with output_file.open("w") as f:
    json.dump(all_vd_nums, f)

print(f"Saved {len(all_vd_nums)} VD numbers to {output_file}")
```

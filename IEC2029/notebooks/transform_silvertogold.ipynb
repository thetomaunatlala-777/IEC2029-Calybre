```python
%pip install pandas
import pandas as pd 
import os
from sqlalchemy import create_engine
from dotenv import load_dotenv, find_dotenv
from pathlib import Path
```

```python
# connect to database
env_path = Path(__file__).parent / ".env" if "__file__" in dir() else Path("../.env")
load_dotenv(dotenv_path=env_path, override=True)

engine = create_engine(
    f"postgresql://{os.getenv('DB_USER')}:{os.getenv('DB_PASSWORD')}"
    f"@{os.getenv('DB_HOST_LOCAL')}:{os.getenv('DB_PORT_LOCAL')}/{os.getenv('DB_NAME')}"
)
```

```python
# read and initialise tables
voting_stations = pd.read_sql("SELECT * FROM silver.\"voting_stations\"", con=engine)
wards = pd.read_sql("SELECT * FROM silver.\"mdb\"", con=engine)
population = pd.read_sql("SELECT * FROM silver.\"population\"", con=engine)
election_results = pd.read_sql("SELECT * FROM silver.\"election_results\"", con=engine)
```
